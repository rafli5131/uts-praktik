## Soal 1 – Batch Processing Pipeline dengan Python

````python
import pandas as pd

def clean_sensor_data(df):
    df.columns = df.columns.str.strip()
    
    df['temperature_kelvin'] = pd.to_numeric(df['temperature_kelvin'], errors='coerce')
    df['humidity_percent'] = pd.to_numeric(df['humidity_percent'], errors='coerce')

    median_temp = df['temperature_kelvin'].median()
    median_humidity = df['humidity_percent'].median()
    
    df['temperature_kelvin'] = df['temperature_kelvin'].fillna(median_temp)
    df['humidity_percent'] = df['humidity_percent'].fillna(median_humidity)
    
    return df

def calculate_aggregates(df):
    aggregated_df = df.groupby('truck_id').agg(
        avg_temperature_kelvin=('temperature_kelvin', 'mean'), 
        max_humidity_percent=('humidity_percent', 'max')
    ).reset_index()
    
    aggregated_df['avg_temperature_kelvin'] = aggregated_df['avg_temperature_kelvin'].round(2)
    aggregated_df['max_humidity_percent'] = aggregated_df['max_humidity_percent'].round(2)
    
    return aggregated_df

df = pd.read_csv('./datasets/sensor_data_raw.csv')
cleaned_df = clean_sensor_data(df)
result_df = calculate_aggregates(cleaned_df)
result_df.to_csv('./datasets/aggregated_results.csv', index=False)
````

clean_sensor_data(df) berfungsi untuk membersihkan data yang kosongfungsi ini bekerja dengan mengambil data dari kolom temperature_kelvin dan humidity_percent, menghitung median/nilai tengah dari masing masing kolom kemudian mengisi nilai kosong dengan median

calculate_aggregates(df) berfungsi menghitung nilai rata rata dari nilai temperature_kelvin dan nilai maksimal dari humidity_percent berdasarkan truck_id.

## Soal 3 – Advanced SQL dengan Window Functions

Import data  datasets/sample_database.sql menggunakan phpmyadmin

Jalankan Procedure untuk memperbanyak data

Query untuk detect temperature anomalies menggunakan LAG() function

```sql
WITH temperature_analysis AS (
	SELECT
		sensor_id,
		temperature AS current_temp,
		recorded_at,
		LAG(temperature) OVER (PARTITION BY sensor_id ORDER BY recorded_at) AS prev_temp
	FROM sensor_readings_hist
),
anomalies AS (
	SELECT
		sensor_id,
		current_temp,
		prev_temp,
		-- Calculate absolute difference; NULLs ignored later
		ABS(current_temp - prev_temp) AS temperature_diff,
		recorded_at,
		CASE
			WHEN prev_temp IS NULL THEN NULL
			WHEN ABS(current_temp - prev_temp) > 20.0 THEN 'HIGH'
			WHEN ABS(current_temp - prev_temp) > 10.0 THEN 'MEDIUM'
			WHEN ABS(current_temp - prev_temp) > 0.0 THEN 'LOW'
			ELSE NULL
		END AS anomaly_level
	FROM temperature_analysis
)
SELECT
	sensor_id,
	current_temp,
	prev_temp,
	temperature_diff,
	anomaly_level,
	recorded_at
FROM anomalies
WHERE temperature_diff IS NOT NULL
	AND temperature_diff > 10.0
ORDER BY temperature_diff DESC, recorded_at DESC;
```

Query di atas menggunakan metode subquery dimana pertama kita harus mencari temperature_analysis menggunakan fungsi LAG() untuk mengetahui temperatur sebelumnya , setelah itu subquery tersebut kita masukkan ke sub query kedua anomalies dimana di sini kita mengklasifikasikan tingkat anomali menjadi LOW,MEDIUM,dan HIGH , Langkah terakhir adalah memfilter subquery anomalies dimana temperature_diff kurang dari 10 derajat.

## Soal 4 – Simple ETL Pipeline

Extract data dari JSON, transform, dan load ke CSV. Implementasi singkat di sel kode berikut.

````python
import pandas as pd
import json
from pathlib import Path

def extract_json_data(file_path):
    """Extract dan flatten JSON data"""
    p = Path(file_path)
    if not p.exists():
        raise FileNotFoundError(f"File not found: {file_path}")
    with p.open('r', encoding='utf-8') as f:
        data = json.load(f)

    # Flatten nested readings array into rows
    rows = []
    for sensor in data:
        sensor_id = sensor.get('sensor_id')
        truck_id = sensor.get('truck_id')
        for r in sensor.get('readings', []):
            row = {
                'sensor_id': sensor_id,
                'truck_id': truck_id,
                'timestamp': r.get('timestamp'),
                'temp_c': r.get('temp'),
                'humidity': r.get('humidity')
            }
            rows.append(row)
    df = pd.DataFrame(rows)
    return df

def transform_data(df):
    """Apply transformations"""
    df = df.copy()
    # Ensure numeric types
    df['temp_c'] = pd.to_numeric(df['temp_c'], errors='coerce')
    df['humidity'] = pd.to_numeric(df['humidity'], errors='coerce')

    # Convert C to F
    df['temp_f'] = (df['temp_c'] * 9/5) + 32

    # comfort_index = temp / humidity * 100 (use temp in C per instructions)
    df['comfort_index'] = (df['temp_c'] / df['humidity']) * 100

    # Round numeric columns for neatness
    df['temp_c'] = df['temp_c'].round(2)
    df['temp_f'] = df['temp_f'].round(2)
    df['comfort_index'] = df['comfort_index'].round(2)
    return df

def load_to_csv(df, output_path):
    """Save to CSV"""
    p = Path(output_path)
    p.parent.mkdir(parents=True, exist_ok=True)
    df.to_csv(p, index=False)
    return p.resolve()

# Execute ETL pipeline using dataset path relative to repository
input_path = 'datasets/sensor_data_raw.csv'  # fallback if sensors.json not present
json_path = 'datasheet/sensors.json'
output_csv = 'datasets/etl_output.csv'

# If the example JSON exists in datasheet, use it; otherwise try fallback
if Path(json_path).exists():
    raw = extract_json_data(json_path)
else:
    # Attempt to read a minimal JSON from working directory as provided in problem statement
    # If not present, raise a helpful error
    raise FileNotFoundError(f"Neither {json_path} found. Please place sensors.json in 'datasheet/'")

transformed = transform_data(raw)
out_path = load_to_csv(transformed, output_csv)
print(f"Wrote ETL output to: {out_path}")
````

## Soal 5 – Data Quality Validation

````python
import pandas as pd

class DataQualityChecker:

    def __init__(self):
        self.violations = []

    def check_missing_values(self, df: pd.DataFrame, required_cols: list):
        for col_name in required_cols:
            # Identifikasi NaN/None
            missing_mask = df[col_name].isnull()
            
            # Identifikasi string kosong
            if df[col_name].dtype == object:
                missing_mask |= df[col_name].astype(str).str.strip().eq("")
            
            missing_count = int(missing_mask.sum())
            
            if missing_count > 0:
                example_missing_values = df.loc[missing_mask, col_name].head(3).tolist()
                
                self.violations.append({
                    'check': 'Missing Values', 
                    'column': col_name, 
                    'count': missing_count,
                    'examples': example_missing_values,
                    'description': f"Ditemukan {missing_count} nilai hilang (NaN/kosong) di kolom {col_name}."
                })

    def check_value_ranges(self, df: pd.DataFrame):
        temp_series = pd.to_numeric(df.get('temp'), errors='coerce')
        humidity_series = pd.to_numeric(df.get('humidity'), errors='coerce')
        
        # --- Pengecekan Temperatur (-40°C hingga 80°C) ---
        invalid_temp_mask = temp_series.notna() & ~temp_series.between(-40, 80)
        invalid_temp_count = int(invalid_temp_mask.sum())
        
        if invalid_temp_count > 0:
            example_temp = temp_series[invalid_temp_mask].head(3).tolist()
            self.violations.append({
                'check': 'Value Range', 
                'column': 'temp', 
                'count': invalid_temp_count,
                'examples': example_temp, 
                'description': f"Ditemukan {invalid_temp_count} nilai suhu di luar rentang [-40, 80]."
            })

        # --- Pengecekan Kelembaban (0% hingga 100%) ---
        invalid_humidity_mask = humidity_series.notna() & ~humidity_series.between(0, 100)
        invalid_humidity_count = int(invalid_humidity_mask.sum())
        
        if invalid_humidity_count > 0:
            example_humidity = humidity_series[invalid_humidity_mask].head(3).tolist()
            self.violations.append({
                'check': 'Value Range', 
                'column': 'humidity', 
                'count': invalid_humidity_count,
                'examples': example_humidity, 
                'description': f"Ditemukan {invalid_humidity_count} nilai kelembaban di luar rentang [0, 100]."
            })

    def check_duplicates(self, df: pd.DataFrame, key_columns: list):
        overall_duplicate_count = int(df.duplicated(subset=key_columns, keep='first').sum())
        
        per_column_duplicate_counts = {}
        per_column_duplicate_examples = {}
        null_placeholder = '__NULL__'
        
        for col_name in key_columns:
            value_counts = df[col_name].fillna(null_placeholder).astype(str).value_counts()
            duplicate_values = value_counts[value_counts > 1]
            count_total_instances = int(duplicate_values.sum()) if not duplicate_values.empty else 0
            per_column_duplicate_counts[col_name] = count_total_instances
            per_column_duplicate_examples[col_name] = [
                {'value': (None if val == null_placeholder else val), 'count': int(value_counts[val])} 
                for val in duplicate_values.index[:3]
            ]
        
        if overall_duplicate_count > 0 or any(per_column_duplicate_counts.values()):
            self.violations.append({
                'check': 'Duplicates', 
                'column': 'Record/Value', 
                'count': overall_duplicate_count,
                'per_column': per_column_duplicate_counts,
                'examples': per_column_duplicate_examples,
                'description': f"Ditemukan Nilai duplikat di data."
            })

    def generate_report(self):
        if not self.violations:
            return "--- Data Quality Report ---\nTidak ada pelanggaran yang ditemukan."
        
        output_lines = ["--- Data Quality Report ---", f"Total Pelanggaran Ditemukan: {len(self.violations)}", ""]
        
        for i, violation in enumerate(self.violations, 1):
            output_lines.append(f"[{i}] JENIS: {violation['check']} | KOLOM: {violation.get('column')} | JUMLAH KASUS: {violation.get('count')}")
            
            if 'per_column' in violation and violation['per_column']:
                output_lines.append("  DETAIL DUPLIKASI PER KOLOM:")
                for col, count in violation['per_column'].items():
                    if count > 0:
                        output_lines.append(f"    - Kolom {col}\t: {count} instance berulang.")
            
            if 'examples' in violation:
                output_lines.append(f"  CONTOH:")
                if isinstance(violation['examples'], dict):
                    for col, examples in violation['examples'].items():
                         if examples:
                             output_lines.append(f"    - Kolom {col}\t: {examples}")
                else:
                    output_lines.append(f"    - Nilai: {violation['examples']}")

            output_lines.append(f"  CATATAN: {violation.get('description')}")
            output_lines.append("")
        return "\n".join(output_lines)

# --- Test Case ---
test_data = pd.DataFrame([
    {"sensor_id": "SENS001", "truck_id": "TRK001", "temp": -999, "humidity": 120},  # Invalid values
    {"sensor_id": "SENS002", "truck_id": "", "temp": 25.5, "humidity": None},       # Missing/Empty values  
    {"sensor_id": "SENS001", "truck_id": "TRK001", "temp": 25.5, "humidity": 65},   # Duplicate
    {"sensor_id": "SENS001", "truck_id": "TRK001", "temp": 25.5, "humidity": 65}    # Duplicate
])

checker = DataQualityChecker()
checker.check_missing_values(test_data.copy(), ['sensor_id', 'truck_id', 'humidity'])
checker.check_value_ranges(test_data.copy())
checker.check_duplicates(test_data.copy(), ['sensor_id', 'truck_id', 'temp', 'humidity'])
report = checker.generate_report()
print(report)
````

check_missing_values(df, required_cols) berfungsi memastikan tidak ada data yang kosong di kolom yang ditentukan (sensor_id, truck_id, humidity). Fungsi ini akan mengidentifikasi nilai yang kosong (NaN/None/” ”). Jika ditemukan nilai kosong, program akan mencatat jumlahnya.
check_value_ranges(df) berfungsi untuk memvalidasi nilai berada pada batas nilai yang telah ditentukan
Temperature: -40 hingga 80 C
Humidity: 0 hingga 100%
check_duplicates(df, key_columns) berfungsi untuk memeriksa duplikasi pada perekaman sensor.
generate_report() berfungsi menggabungkan semua pelanggaran dalam array self.violations menjadi output string yang terstruktur dan mudah dibaca.
